 # Handle Missing Values

```python
# check for null values
print(df.isnull().sum())
```

```python
# filling missing values
df = df.drop(columns=['videos', 'seller_information'])

# fill the ones that make sense
df['discount'] = df['discount'].fillna(df['discount'].median())
df['what_customers_said'] = df['what_customers_said'].fillna('No feedback')
df['seller_name'] = df['seller_name'].fillna('Unknown')
df['variations'] = df['variations'].fillna('No variations')

print(df.isnull().sum())
```

# Basic Operations (Filter & Select)

```python
# products with rating more than 4
high_rated = df[df['rating'] > 4]
print(high_rated.shape)
high_rated.head()
```

```python
# selecting only useful columns
df[['title', 'category', 'rating', 'initial_price', 'discount']].head()
```

# Remove Duplicates

```python
print("before:", len(df))
df = df.drop_duplicates()
print("after:", len(df))
```

# Derived Column (total_amount)

```python
# dataset doesnt have quantity column so i added it manually as 1
df['quantity'] = 1
df['total_amount'] = df['initial_price'] * df['quantity']

df[['title', 'initial_price', 'quantity', 'total_amount']].head()
```

#  Save Cleaned CSV

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
